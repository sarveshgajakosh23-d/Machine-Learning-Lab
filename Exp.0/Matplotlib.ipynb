# 🌟 Mini Project

**Student Performance Dashboard**

<img src="https://matplotlib.org/stable/_static/logo2.svg" width="90">

```text
Data → Matplotlib → Graph → Insights
```

💡 **Did You Know?**
- Good visualizations make complex data easier to understand.

📝 **Practice Exercise**
- Change the values.
- Try a different chart type.
- Predict the output before running.


````python
import matplotlib.pyplot as plt

students=["A","B","C","D","E"]
marks=[78,91,66,88,95]

plt.figure(figsize=(6,4))
plt.bar(students,marks)
plt.title("Student Performance Dashboard")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.grid(axis='y')
plt.show()
````
