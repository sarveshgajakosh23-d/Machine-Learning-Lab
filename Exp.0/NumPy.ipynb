# 🌟 Visualization

**Line, Bar, Scatter, Histogram, Pie**

<img src="https://upload.wikimedia.org/wikipedia/commons/3/31/NumPy_logo_2020.svg" width="90">

### 📌 Concept
Explain the topic in simple language.

```text
Input → NumPy Processing → Result
```

💡 **Did You Know?**
- NumPy uses vectorized operations which are much faster than Python loops.

📝 **Practice Exercise**
1. Try modifying the values.

````python
import numpy as np
import matplotlib.pyplot as plt
x=np.arange(1,6)
y=x**2
plt.plot(x,y,marker='o')
plt.grid()
plt.show()

plt.scatter(x,y)
plt.show()

plt.hist(np.random.randn(500),bins=20)
plt.show()

plt.pie([40,30,20,10],labels=["A","B","C","D"],autopct="%1.1f%%")
plt.show()
````
