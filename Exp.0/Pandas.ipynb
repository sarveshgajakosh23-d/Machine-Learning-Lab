# 🌟 Visualization

**Line, Bar, Histogram, Pie**

<img src="https://pandas.pydata.org/static/img/pandas_mark.svg" width="80">

```text
Raw Data → Pandas → Analysis → Insights
```

💡 **Did You Know?**
- Pandas DataFrames are like Excel spreadsheets in Python.

📝 **Practice**
- Modify the dataset.
- Predict the output before running.


````python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.DataFrame({
    "Student":["A","B","C","D"],
    "Marks":[80,92,75,88]
})

df.plot(x="Student",y="Marks",kind="bar",legend=False,title="Marks")
plt.show()

df.plot(x="Student",y="Marks",kind="line",marker="o")
plt.show()

df["Marks"].plot(kind="hist",bins=5)
plt.show()

df.set_index("Student")["Marks"].plot(kind="pie",autopct="%1.1f%%")
plt.ylabel("")
plt.show()
````

# 🌟 Mini Project

**Student Performance Analyzer**

<img src="https://pandas.pydata.org/static/img/pandas_mark.svg" width="80">

```text
Raw Data → Pandas → Analysis → Insights
```

💡 **Did You Know?**
- Pandas DataFrames are like Excel spreadsheets in Python.

📝 **Practice**
- Modify the dataset.
- Predict the output before running.


````python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.DataFrame({
"Student":["A","B","C","D","E"],
"Marks":[85,92,76,88,95]
})

print("Average:",df["Marks"].mean())
print("Highest:",df["Marks"].max())
print("Lowest:",df["Marks"].min())

df.plot(x="Student",y="Marks",kind="bar",legend=False,title="Student Performance")
plt.show()
````
