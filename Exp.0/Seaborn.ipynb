# 🌟 Line & Bar Plots

**Visualize trends and comparisons**

```text
Dataset → Seaborn → Visualization → Insights
```

💡 **Did You Know?**
- Seaborn automatically applies attractive themes and color palettes.

📝 **Practice Exercise**
- Modify the dataset.
- Change the plot style.
- Predict the output before running.


````
import seaborn as sns
import matplotlib.pyplot as plt

tips=sns.load_dataset("tips")

sns.lineplot(data=tips,x="size",y="total_bill")
plt.show()

sns.barplot(data=tips,x="day",y="total_bill")
plt.show()
````

# 🌟 Distribution Plots

**Histogram, KDE, Box Plot**

```text
Dataset → Seaborn → Visualization → Insights
```

💡 **Did You Know?**
- Seaborn automatically applies attractive themes and color palettes.

📝 **Practice Exercise**
- Modify the dataset.

````
import seaborn as sns
import matplotlib.pyplot as plt

tips=sns.load_dataset("tips")

sns.histplot(tips["total_bill"],kde=True)
plt.show()

sns.boxplot(data=tips,x="day",y="total_bill")
plt.show()
````

# 🌟 Heatmaps

**Correlation Matrix**

```text
Dataset → Seaborn → Visualization → Insights
```

💡 **Did You Know?**
- Seaborn automatically applies attractive themes and color palettes.

📝 **Practice Exercise**
- Modify the dataset.
- Change the plot style.
- Predict the output before running.


````
import seaborn as sns
import matplotlib.pyplot as plt

iris=sns.load_dataset("iris")
corr=iris.corr(numeric_only=True)

sns.heatmap(corr,annot=True,cmap="Blues")
plt.show()
````

# 🌟 Mini Project

**Student Performance Analysis**

```text
Dataset → Seaborn → Visualization → Insights
```

💡 **Did You Know?**
- Seaborn automatically applies attractive themes and color palettes.

📝 **Practice Exercise**
- Modify the dataset.

- Predict the output before running.


````
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

df=pd.DataFrame({
"Student":["A","B","C","D","E"],
"Marks":[85,91,76,88,95]
})

sns.barplot(data=df,x="Student",y="Marks")
plt.title("Student Performance Analysis")
plt.show()
````
